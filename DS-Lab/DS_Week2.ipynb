```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age':[25,30,np.nan,40,35],
    'Department':['HR','Finance','Finance',np.nan,'IT']})
print(df)

#mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

#mode
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age':[25,30,np.nan,40,35],
    'Department':['HR','Finance','Finance',np.nan,'IT']})
print("Original dataset with null values")
print(df)
#Forward fill
df_ffill = df.copy()
df_ffill.ffill(inplace=True)
print("Forward fill")
print(df_ffill)
#Backward fill
df_bfill = df.copy()
df_bfill.bfill(inplace = True)
print("Backward fill")
print(df_bfill)

```

```python
#drop rows with missing values
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'Age' : [34,67, np.nan ,99,10],
    'Department' : ['Finance','Finance',np.nan,'HR','IT']
})
print("Original Dataset")
print(df)
df_drop_row = df.dropna(axis=0)
print("Dataset after drop row")
print(df_drop_row)
#drop column with missing value
df_drop_cols = df.dropna(axis=1)#if axis = 0 it removes row with missing value
print("Dataset after dropping columns with missing values:\n",df_drop_cols)
```

```python
#removal of the duplicates
import pandas as pd
df = pd.DataFrame({
    'ID':[1,2,2,3,4,4],
    'Name':['Alice','Bob',"Bob",'bob','maria',"Charlie"],
    'Age':[25,30,30,35,40,40]
})
print("Original Data:\n",df)
df_exact = df.drop_duplicates()
print("DataSet after deleting duplicates\n",df_exact)
```

```python
#removing a subset
import pandas as pd
df = pd.DataFrame({
    'ID':[1,2,2,3,4,4],
    'Name':['Alice','Bob',"Bob",'bob','maria',"Charlie"],
    'Age':[25,30,30,35,40,40]
})
print("Original Data:\n",df)
df_exact = df.drop_duplicates(subset=['Age'])
print("DataSet after deleting duplicates\n",df_exact)
df_exact = df.drop_duplicates(subset=['ID'])
print("DataSet after deleting duplicates\n",df_exact)
df_exact = df.drop_duplicates(subset=['Name'])
print("DataSet after deleting duplicates\n",df_exact)
```

```python
#correcting inconsistent date format
import pandas as pd
df = pd.DataFrame({
    'Date':['2025-01-05','05/01/2025','Jan 5, 2025','2025.01.05']
})
print(df)

df['Date'] = pd.to_datetime(df['Date'],errors='coerce').dt.strftime('%Y-%m-%d')

print(df)
```

```python
#case normalization
import pandas as pd
df = pd.DataFrame({
    'Name':['Alice','DAVID','charlie','BoB']
})
df['lower'] = df['Name'].str.lower()
df['upper'] = df['Name'].str.upper()
df['Caplita'] = df['Name'].str.capitalize()
print(df)
```
